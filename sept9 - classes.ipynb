```python
from texthandler import TextProcessor
```

```python
from filehandler import FileHandler
```

```python


# Instantiate the FileHandler to read a file
file_handler = FileHandler('transcript.txt')
transcript = file_handler.read_file()

# Instantiate the TextProcessor to process the text
processor = TextProcessor(transcript)

# Use the class methods
summary = processor.summarize(max_length=10)
word_count = processor.word_count()

# Output the results
print("Summary:", summary)
print("Word Count:", word_count)

# Write the summary back to a file
file_handler.write_file(summary)
```
